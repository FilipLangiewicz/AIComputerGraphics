```python
import os
import trimesh

import numpy as np

from pathlib import Path

from bunny.utils import load_mesh, visualize_transition, visualize_transition_3d
```

```python
PROJ_ROOT = Path(os.getcwd()).parents[0]
MODELS_DIR = PROJ_ROOT / "modele"
BUNNY_OBJECT_PATH = MODELS_DIR / "bunny.obj"
TEAPOT_OBJECT_PATH = MODELS_DIR / "teapot.obj"

print(f"Project root: {PROJ_ROOT} (exists: {PROJ_ROOT.exists()}) \n")
print(f"Models dir: {MODELS_DIR} (exists: {MODELS_DIR.exists()}) \n")
print(f"Bunny object: {BUNNY_OBJECT_PATH} (exists: {BUNNY_OBJECT_PATH.exists()}) \n")
print(f"Teapot object: {TEAPOT_OBJECT_PATH} (exists: {TEAPOT_OBJECT_PATH.exists()}) \n")
```

```python
bunny = load_mesh(BUNNY_OBJECT_PATH)
teapot = load_mesh(TEAPOT_OBJECT_PATH)

print(f"Bunny — vertices: {len(bunny.vertices)}, faces: {len(bunny.faces)}")
print(f"Teapot — vertices: {len(teapot.vertices)}, faces: {len(teapot.faces)}")
```

```python
n_points = 10000

bunny_points, _ = trimesh.sample.sample_surface(bunny, n_points)
teapot_points, _ = trimesh.sample.sample_surface(teapot, n_points)
```

```python
visualize_transition(
    [bunny_points, teapot_points],
    titles=["Bunny", "Teapot"]
)
```

```python
visualize_transition_3d(
    [bunny_points, teapot_points],
    titles=["Bunny", "Teapot"]
)
```

```python
for name, pts in [("Bunny", bunny_points), ("Teapot", teapot_points)]:
    print(f"{name} | center: {pts.mean(axis=0).round(3)} | scale: {np.linalg.norm(pts, axis=1).max():.3f}")
```
