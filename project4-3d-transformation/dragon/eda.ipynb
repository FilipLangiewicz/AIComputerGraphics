```python
import os
import trimesh

import numpy as np

from pathlib import Path

from dragon.utils import load_mesh, visualize_transition, visualize_transition_3d
```

```python
PROJ_ROOT = Path(os.getcwd()).parents[0]
MODELS_DIR = PROJ_ROOT / "modele"
DRAGON_OBJECT_PATH = MODELS_DIR / "dragon_small.obj"
TEAPOT_OBJECT_PATH = MODELS_DIR / "teapot.obj"

print(f"Project root: {PROJ_ROOT} (exists: {PROJ_ROOT.exists()}) \n")
print(f"Models dir: {MODELS_DIR} (exists: {MODELS_DIR.exists()}) \n")
print(f"Dragon object: {DRAGON_OBJECT_PATH} (exists: {DRAGON_OBJECT_PATH.exists()}) \n")
print(f"Teapot object: {TEAPOT_OBJECT_PATH} (exists: {TEAPOT_OBJECT_PATH.exists()}) \n")
```

```python
dragon = load_mesh(DRAGON_OBJECT_PATH)
teapot = load_mesh(TEAPOT_OBJECT_PATH)

print(f"Dragon — vertices: {len(dragon.vertices)}, faces: {len(dragon.faces)}")
print(f"Teapot — vertices: {len(teapot.vertices)}, faces: {len(teapot.faces)}")
```

```python
n_points = 10000

dragon_points, _ = trimesh.sample.sample_surface(dragon, n_points)
teapot_points, _ = trimesh.sample.sample_surface(teapot, n_points)
```

```python
visualize_transition(
    [dragon_points, teapot_points],
    titles=["Dragon", "Teapot"]
)
```

```python
visualize_transition_3d(
    [dragon_points, teapot_points],
    titles=["Dragon", "Teapot"]
)
```

```python
for name, pts in [("Dragon", dragon_points), ("Teapot", teapot_points)]:
    print(f"{name} | center: {pts.mean(axis=0).round(3)} | scale: {np.linalg.norm(pts, axis=1).max():.3f}")
```
