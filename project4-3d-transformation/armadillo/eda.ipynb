```python
import sys
sys.path.insert(0, "..")

from armadillo.utils import load_mesh, visualize_transition
import trimesh
import numpy as np
```

```python
armadillo = load_mesh("../modele/armadillo_small.obj")
teapot    = load_mesh("../modele/teapot.obj")

print(f"Armadillo — vertices: {len(armadillo.vertices)}, faces: {len(armadillo.faces)}")
print(f"Teapot    — vertices: {len(teapot.vertices)}, faces: {len(teapot.faces)}")
```

```python
n_points = 10000
arm_pts, _ = trimesh.sample.sample_surface(armadillo, n_points)
tea_pts, _ = trimesh.sample.sample_surface(teapot, n_points)
```

```python
visualize_transition(
    [arm_pts, tea_pts],
    titles=["Armadillo", "Teapot"]
)
```

```python
from armadillo.utils import visualize_transition_3d

visualize_transition_3d(
    [arm_pts, tea_pts],
    titles=["Armadillo", "Teapot"]
)
```

```python
for name, pts in [("Armadillo", arm_pts), ("Teapot", tea_pts)]:
    print(f"{name} | center: {pts.mean(axis=0).round(3)} | scale: {np.linalg.norm(pts, axis=1).max():.3f}")
```
